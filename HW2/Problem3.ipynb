```python
# input: 
# n (network size), gamma (powerlaw exponent). and <k> 
def powerlaw_configmodel(n, gamma, k_mean):
    # step 1: set k_min and k_max
    k_min = 1
    k_max = int(np.sqrt(n))
    
    # 2. Sample from continuous power-law and discretize
    # Using inverse transform sampling
    rng = np.random.default_rng(42)
    r = rng.random(n)
    a = gamma - 1.0
    k_vals = ((k_max**a - k_min**a) * r + k_min**a) ** (1.0 / a)
    
    # 3. Normalize and scale to desired mean degree
    k_vals *= k_mean / np.mean(k_vals)
    k_vals = np.round(k_vals).astype(int)
    
    # 4. fix the bound 
    k_vals[k_vals < k_min] = k_min
    k_vals[k_vals > k_max] = k_max
    
    # 5. makesure sum of k is even
    if np.sum(k_vals) % 2 != 0:
        k_vals[0] += 1
    
        
    return k_vals
```

```python
n = 1000
gamma = 2.5
k_mean = 20

k_vals = powerlaw_configmodel(n, gamma, k_mean)
print(k_vals)
```

```python
# input
# degrees list of k_i, mu, communities (list specifying each node's community label)
# output
# two arrays: k_in[i] = internal degree of node i and k_out[i] = external degree of node i

def assign_mu(mu, k_vals):
    k_in = [0] * len(k_vals)
    k_out = [0] * len(k_vals)

    k_vals = np.asarray(k_vals, dtype=int)
    k_in = np.rint((1-mu) * k_vals).astype(int)
    k_out = k_vals - k_in    
    
    k_in = k_in.tolist()
    k_out = k_out.tolist()

    return k_in, k_out
```

```python
mu = 0.5
k_in, k_out = assign_mu(mu, k_vals)
```
